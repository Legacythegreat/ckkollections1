```python
# Cleanup Conflicting Files
This notebook scans the workspace for common conflicting file types, performs a safe cleanup, and verifies the result.
</VSCode.Cell>
<VSCode.Cell id="2" language="python">
import os
from pathlib import Path
import shutil

project_root = Path(r'c:\Users\SCL\Desktop\ck kollections')
print(f'Project root set to: {project_root}')
</VSCode.Cell>
<VSCode.Cell id="3" language="markdown">
## Identify Conflicting File Types
Define file patterns and extensions that are common sources of conflict in mixed web app projects.
</VSCode.Cell>
<VSCode.Cell id="4" language="python">
conflict_patterns = [
    '*.log',
    '*.tmp',
    '*.temp',
    '*.bak',
    '*.swp',
    '*.swo',
    '__pycache__',
    '*.pyc',
    '*.pyo',
    '*.class',
    '*.jar',
    '*.DS_Store',
    '.DS_Store',
    'Thumbs.db',
    'node_modules',
    'dist',
    'build',
    '.vite',
    '.cache',
    'yarn.lock',
    'package-lock.json',
    'pnpm-lock.yaml',
    'npm-debug.log',
    'yarn-error.log',
    '.env',
    '.env.local',
    '.env.development',
    '.env.production',
    'package.json',
    'vite.config.js',
    'tailwind.config.js',
    'postcss.config.js',
    'README.md',
    'index.html',
    'src',
    'backend',
    'php-version',
    'cleanup.cmd',
    '.gitignore',
]

print('Conflict patterns defined.')
</VSCode.Cell>
<VSCode.Cell id="5" language="markdown">
## Scan Directory for Files to Remove
Recursively search for matching files and directories.
</VSCode.Cell>
<VSCode.Cell id="6" language="python">
def matches(path: Path) -> bool:
    name = path.name
    if name in conflict_patterns:
        return True
    return any(path.match(pattern) for pattern in conflict_patterns)

candidates = []
for root, dirs, files in os.walk(project_root):
    root_path = Path(root)
    for d in dirs:
        p = root_path / d
        if matches(p):
            candidates.append(p)
    for f in files:
        p = root_path / f
        if matches(p):
            candidates.append(p)

print(f'Found {len(candidates)} candidate(s) for removal')
for path in candidates:
    print(path)
</VSCode.Cell>
<VSCode.Cell id="7" language="markdown">
## Remove Files Safely
Delete the identified file and directory candidates after confirmation.
</VSCode.Cell>
<VSCode.Cell id="8" language="python">
dry_run = True

if dry_run:
    print('Dry run enabled. No files will be removed.')
    for path in candidates:
        print(path)
else:
    for path in sorted(candidates, key=lambda p: str(p), reverse=True):
        if path.is_dir():
            shutil.rmtree(path)
            print(f'Removed directory: {path}')
        elif path.is_file() or path.is_symlink():
            path.unlink()
            print(f'Removed file: {path}')

print('Removal step complete.')
</VSCode.Cell>
<VSCode.Cell id="9" language="markdown">
## Verify Cleanup
Scan the workspace again and report the final state after cleanup.
</VSCode.Cell>
<VSCode.Cell id="10" language="python">
remaining = []
for root, dirs, files in os.walk(project_root):
    root_path = Path(root)
    for d in dirs:
        p = root_path / d
        if matches(p):
            remaining.append(p)
    for f in files:
        p = root_path / f
        if matches(p):
            remaining.append(p)

print(f'Remaining conflicting paths: {len(remaining)}')
for path in remaining:
    print(path)
```
